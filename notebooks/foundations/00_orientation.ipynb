## 3. The one piece of math every notebook after this one assumes
Almost everything in this curriculum reduces to one operation: a
**quadratic form**, `w' Sigma w` — "w transpose Sigma w." If matrix
notation isn't yet second nature, work through this by hand once here so
it's automatic later.

**Vectors and dot products.** A portfolio's weights are a vector,
`w = [w_1, w_2, ...]`. The dot product of two vectors multiplies matching
entries and adds them up: `a . b = sum_i(a_i * b_i)`. A portfolio's
expected return is a dot product: `E[r_p] = w . mu = sum_i(w_i * mu_i)`.

**Matrix-vector products.** A covariance matrix `Sigma` is a grid: entry
`Sigma[i][j]` is the covariance between asset `i` and asset `j` (and
`Sigma[i][i]` is asset `i`'s own variance). `Sigma @ w` (matrix times
vector) produces a new vector, one entry per asset, where entry `i` is
the dot product of `Sigma`'s row `i` with `w`.

**The quadratic form.** `w' Sigma w` means: first compute `Sigma @ w`
(a vector), then dot that result with `w` again (a scalar). This scalar
*is* portfolio variance — `sigma_p^2 = w' Sigma w`.

## Why this matters
Every risk formula in this repo is a variation on this one operation:
`portfolio_variance` (`w' Sigma w` directly), `tracking_error`
(the same thing applied to active weights `w_p - w_b`), `marginal_risk_contribution`
(`(Sigma @ w) / sigma_p` — literally step 1 above, divided by the
volatility), `factor_model_covariance` (`B F B' + D`, the same
matrix-multiplication pattern one layer deeper). None of it is new math
once this one operation is automatic — it's the same few lines of linear
algebra, applied to different inputs.
